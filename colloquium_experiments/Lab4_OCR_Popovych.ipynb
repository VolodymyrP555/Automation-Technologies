```python

import cv2
import easyocr
import matplotlib.pyplot as plt
import os

images = ["chek11.jpg", "check22.jpg", "check33.jpg"]

print("Завантаження моделі EasyOCR...")
reader = easyocr.Reader(['uk', 'en'], gpu=False)

for filename in images:
    print(f"\n{'=' * 45}")
    print(f"Обробка: {filename}")

    img = cv2.imread(filename)

    if img is None:
        print(f"Помилка: не вдалося відкрити {filename}")
        continue

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    results = reader.readtext(gray)

    output = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    extracted_text = []

    for bbox, text, confidence in results:
        x1, y1 = map(int, bbox[0])
        x2, y2 = map(int, bbox[2])

        cv2.rectangle(output, (x1, y1), (x2, y2), (0, 255, 0), 2)

        if confidence > 0.3:
            extracted_text.append(text)
            print(f"Текст: {text} | Впевненість: {confidence:.2f}")

    txt_name = filename.replace(".jpg", "_ocr.txt")

    with open(txt_name, "w", encoding="utf-8") as f:
        f.write("\n".join(extracted_text))

    print(f"Знайдено блоків: {len(results)}")
    print(f"Збережено: {txt_name}")

    plt.figure(figsize=(9, 11))
    plt.imshow(output)
    plt.title(f"OCR: {filename}")
    plt.axis("off")
    plt.show()

print("\nОбробку всіх фотографій завершено!")
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("chek11.jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

def evaluate_ocr(img):
    results = reader.readtext(img)
    avg_conf = (
        sum(conf for _, _, conf in results) / len(results)
        if results else 0
    )
    words = sum(len(text.split()) for _, text, conf in results if conf > 0.3)
    return avg_conf, words

base_conf, base_words = evaluate_ocr(gray)

print(f"Оригінал: confidence={base_conf:.3f}, слів={base_words}")

history = [(0.0, base_conf, base_words)]
h, w = gray.shape

for step in range(1, 31):
    shear = round(step * 0.1, 1)

    matrix = np.float32([
        [1, shear, 0],
        [0, 1, 0]
    ])

    new_width = int(w + shear * h)

    distorted = cv2.warpAffine(
        gray, matrix, (new_width, h),
        borderValue=255
    )

    avg_conf, words = evaluate_ocr(distorted)
    history.append((shear, avg_conf, words))

    print(f"Shear={shear:.1f} | Confidence={avg_conf:.3f} | Слів={words}")


    if avg_conf < 0.4 or words <= base_words * 0.5:
        print(f"\nДосягнуто порогу при Shear X = {shear:.1f}")
        break

shears = [x[0] for x in history]
confidences = [x[1] for x in history]
word_counts = [x[2] for x in history]

plt.figure(figsize=(10, 5))
plt.plot(shears, confidences, marker="o", label="Average Confidence")
plt.axhline(0.4, color="red", linestyle="--", label="Поріг 0.4")
plt.xlabel("Коефіцієнт Shear X")
plt.ylabel("Середня впевненість OCR")
plt.title("Вплив горизонтального зсуву на якість OCR")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(shears, word_counts, marker="o", label="Розпізнані слова")
plt.axhline(base_words * 0.5, color="red", linestyle="--",
            label="50% від початкової кількості")
plt.xlabel("Коефіцієнт Shear X")
plt.ylabel("Кількість розпізнаних слів")
plt.title("Вплив горизонтального зсуву на кількість слів")
plt.grid(True)
plt.legend()
plt.show()
```

```python

import cv2
import numpy as np

image = cv2.imread("chek11.jpg")
if image is None:
    raise FileNotFoundError("Завантаж chek11.jpg у Colab")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

h, w = gray.shape
matrix = np.float32([[1, 0.5, 0], [0, 1, 0]])

degraded = cv2.warpAffine(
    gray, matrix, (int(w + 0.5 * h), h),
    borderValue=255
)

reader = easyocr.Reader(["uk", "en"], gpu=False)

clean_results = reader.readtext(gray, detail=0)
degraded_results = reader.readtext(degraded, detail=0)

with open("ocr_result_clean.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(clean_results))

with open("ocr_result_degraded.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(degraded_results))

print("Створено ocr_result_clean.txt")
print("Створено ocr_result_degraded.txt")
print("Чистий текст:", len(clean_results), "блоків")
print("Спотворений текст:", len(degraded_results), "блоків")
```

```python

import cv2
import numpy as np

image = cv2.imread("chek11.jpg")

if image is None:
    raise FileNotFoundError("Завантаж файл chek11(2).jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

def test_ocr(img, threshold=0.3):
    results = reader.readtext(img, detail=1)

    confidence = (
        sum(conf for _, _, conf in results) / len(results)
        if results else 0
    )

    accepted = [
        text for _, text, conf in results
        if conf > threshold
    ]

    return confidence, len(accepted), " ".join(accepted)


h, w = gray.shape

matrix = np.float32([
    [1, 1.5, 0],
    [0, 1, 0]
])

distorted = cv2.warpAffine(
    gray,
    matrix,
    (int(w + 1.5 * h), h),
    borderValue=255
)

experiments = [
    ("Оригінал", gray, 0.3),
    ("Shear X = 1.5", distorted, 0.3),
    ("Поріг 0.75", gray, 0.75),
]

for name, img, threshold in experiments:
    conf, blocks, text = test_ocr(img, threshold)

    print(f"\n=== {name} ===")
    print(f"Confidence: {conf:.4f}")
    print(f"Прийнято блоків: {blocks}")
    print(f"Поріг фільтрації: {threshold}")
    print(f"Розпізнаний текст:\n{text[:1200]}")

    filename = (
        name.replace(" ", "_").replace("=", "")
        + "_result.txt"
    )

    with open(filename, "w", encoding="utf-8") as f:
        f.write(text)
```
